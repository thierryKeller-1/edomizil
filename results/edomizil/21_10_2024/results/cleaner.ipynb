```python
import pandas as pd 
import os
import csv
from datetime import datetime, timedelta
```

```python
date_scrap = "21/10/2024"
```

```python
all_files = os.listdir()    
csv_files = list(filter(lambda f: f.endswith('.csv'), all_files))
dfs = []

for csv_file in csv_files:
    dfs.append(pd.read_csv(csv_file))

len(dfs)
```

```python
df = pd.concat(dfs)
```

```python
df['date_scrap'] = date_scrap
```

```python
df.drop_duplicates(inplace=True, subset=['date_debut', 'date_fin', 'price', 'identifiant',
       'typologie', 'nom'])
len(df)
```

```python
df['date_debut'] = pd.to_datetime(df['date_debut'], format='%d/%m/%Y')
df = df.sort_values(by=['date_debut'])
df['date_debut'] = df['date_debut'].dt.strftime('%d/%m/%Y')
```

```python
df.to_csv(f"edomizil_cleaned_{date_scrap.replace('/', '_')}.csv", index=False)
```
